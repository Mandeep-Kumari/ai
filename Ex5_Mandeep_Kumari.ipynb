```python
import gradio as gr
```

```python

def chatbot_response(message, history):
    text = message.lower().strip()

    if any(word in text for word in ["hello", "hi", "hey"]):
        return "Hello! Welcome to Smart AI Chatbot. How can I help you?"

    elif "nic" in text or "national informatics centre" in text:
        return (
            "NIC stands for National Informatics Centre. It is an organization "
            "under the Ministry of Electronics and Information Technology "
            "(MeitY) that builds and maintains IT infrastructure, networks, "
            "and e-governance services for central, state, and district administrations."
        )

    elif "artificial intelligence" in text or text == "ai":
        return (
            "Artificial Intelligence (AI) enables computers to perform "
            "tasks that normally require human intelligence."
        )

    elif "python" in text:
        return (
            "Python is a popular programming language used in AI, "
            "machine learning, data science, and web development."
        )

    elif "gradio" in text:
        return (
            "Gradio is a Python library used to create interactive "
            "web interfaces for AI and machine learning applications."
        )

    elif "thank" in text:
        return "You're welcome! Feel free to ask me another question."

    else:
        return (
            "I am Smart AI Chatbot. Please ask me about NIC, "
            "Artificial Intelligence, Python, or Gradio."
        )


demo = gr.ChatInterface(
    fn=chatbot_response,
    title="🤖 Smart AI Chatbot",
    description=(
        "Welcome! Ask me about NIC, Artificial Intelligence, "
        "Python, or Gradio."
    ),
    examples=[
        "What is NIC?",
        "What is Artificial Intelligence?",
        "How is Python used in AI?",
        "What is Gradio?"
    ]
)

demo.launch()
```

```python
demo = gr.ChatInterface(
    fn=chatbot_response,
    title="🤖 Smart AI Chatbot",
    description=(
        "Welcome! I am Smart AI Chatbot. Ask me about NIC, Artificial "
        "Intelligence, Python, or Gradio. You can also select a sample prompt below."
    ),
    examples=[
        "What is NIC?",
        "What is Artificial Intelligence?",
        "How is Python used in AI?",
        "What is Gradio?"
    ],
    textbox=gr.Textbox(
        placeholder="Type your message here and press Enter...",
        container=False
    )
)

if __name__ == "__main__":
    demo.launch()
```
