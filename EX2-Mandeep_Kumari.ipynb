```python

import cv2
import numpy as np
```

```python
img1 = cv2.imread(r"C:\Users\hlbs\Desktop\SunflowerPic2.jpg")
img2 = cv2.imread(r"C:\Users\hlbs\Desktop\greenpic1.png")
```

```python
if img1 is None or img2 is None:
  raise FileNotFoundError(
      "Check the file paths! One or both images could not be loaded."
  )
```

```python
# Ensure both images have matching dimensions (width, height)
# OpenCV's .shape returns (height, width, channels)
height, width = img1.shape[:2]
if img2.shape[:2] != (height, width):
  img2 = cv2.resize(img2, (width, height), interpolation=cv2.INTER_AREA)
```

```python
# 2. Perform weighted image blending
# Formula: dst = alpha * img1 + beta * img2 + gamma
alpha = 0.6  # Weight for the first image (60%)
beta = 0.4  # Weight for the second image (40%)
gamma = 0.0  # Scalar added to each sum (brightness offset)

blended = cv2.addWeighted(img1, alpha, img2, beta, gamma)
```

```python
# 3. Display and save the original inputs alongside the blended output
# Concatenate images horizontally (Left: img1, Middle: img2, Right: blended)
comparison = np.hstack((img1, img2, blended))

# Save the individual outputs and the side-by-side comparison for verification
cv2.imwrite("verified_input1.jpg", img1)
cv2.imwrite("verified_input2.jpg", img2)
cv2.imwrite("blended_output.jpg", blended)
cv2.imwrite("side_by_side_verification.jpg", comparison)

print("Images saved successfully!")

# Display the combined result in a window (press any key to close)
cv2.imshow("Original 1 | Original 2 | Blended Output", comparison)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
